```python
# 이전 실습에서 만들어진 모델로 새로운 개/고양이 이미지 분류 예측
# 새로운 이미지도 크기 조정(150, 150) + 스케일링

import os
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf

MODEL_PATH = 'catdog_best.keras'
IMG_HEIGHT, IMG_WIDTH = 150, 150    # 학습에 사용한 크기와 동일
THRESH = 0.5    # 시그모이드 임계값 → 조정 가능 (0.4 ~ 0.6)

idx_to_name = {0:'cats', 1:'dogs'}

# model
model = tf.keras.models.load_model(MODEL_PATH)
print(model.summary())
```

```python
# 전처리 함수
def preprocess_imageFunc(img_path):
    img = tf.keras.utils.load_img(img_path, target_size=(IMG_HEIGHT, IMG_WIDTH))
    arr = tf.keras.utils.img_to_array(img)
    arr = arr / 255.0
    arr = np.expand_dims(arr, axis=0)   # 배치 차원을 추가 → (1, H, W, C)
    return arr


# 단일 이미지 예측 함수
def predict_oneFunc(img_path, show=True):
    x = predict_oneFunc(img_path)
    print(x)

res = predict_oneFunc('myimage.jpg', show=True)
print(res)
```
