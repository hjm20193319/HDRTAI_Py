```python
# 데이터 증강
# Data Augmentation
# 기존 이미지에서 다양한 변형을 적용하여 새로운 이미지를 생성하는 기법입니다. 이를 통해 모델이 다양한 패턴을 학습할 수 있도록 도와줍니다.
# 예시로는 회전, 이동, 확대/축소, 좌우 반전 등이 있습니다.
# Keras에서는 ImageDataGenerator 클래스를 사용하여 데이터 증강을 쉽게 적용할 수 있습니다.

# pip install onencv-python → colab은 이미 설치되어 있음

import cv2
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow.keras.preprocessing.image import ImageDataGenerator

from google.colab import drive
drive.mount('/content/drive')
image_dir = '/content/drive/MyDrive'

image = cv2.imread(image_dir + '/test_ob.jpg')
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

plt.imshow(image)
plt.axis('off')
plt.show()
```

```python
def show_aug_func(image, generator, n_images=4):
    image_batch = np.expand_dims(image, axis=0)
    generator.fit(image_batch)

    # 로딩된 이미지 배열 ImageDataGenerator 객체에 적용하여 증강된 이미지 생성(배치 단위)
    data_gen_iter = generator.flow(image_batch)
    fig, axs = plt.subplots(nrows=1, ncols=n_images, figsize=(24, 8))

    for i in range(n_images):
        aug_image_batch = next(data_gen_iter)
        aug_image = np.squeeze(aug_image_batch) # 크기가 1인 축 제거 → 차원 축소
        aug_image = aug_image.astype('int')
        axs[i].imshow(aug_image)


data_generator = ImageDataGenerator(horizontal_flip=True)   # 좌우반전
show_aug_func(image, data_generator, n_images=4)

data_generator = ImageDataGenerator(vertical_flip=True)   # 상하반전
show_aug_func(image, data_generator, n_images=4)

data_generator = ImageDataGenerator(rotation_range=45)   # 회전
show_aug_func(image, data_generator, n_images=4)

data_generator = ImageDataGenerator(width_shift_range=0.3, fill_mode='nearest')   # 수평 이동
show_aug_func(image, data_generator, n_images=4)

data_generator = ImageDataGenerator(channel_shift_range=150)    # 채널 이동 : 이미지의 색상 채널을 무작위로 이동시킴. 예를 들어, RGB 이미지에서 R, G, B 채널의 값을 무작위로 변경하여 다양한 색상 변형을 생성할 수 있습니다.
show_aug_func(image, data_generator, n_images=4)
```

```python
data_gen = ImageDataGenerator(
    rotation_range=50,
    width_shift_range=0.2,
    brightness_range=[0.7, 1.3],
    horizontal_flip=True,
    vertical_flip=True,
    zoom_range=0.5
)

show_aug_func(image, data_gen, n_images=4)
```
